```r
distribution_reg_wins <- tidy_mm_data |>
        ggplot(aes(x=reg_wins)) +
        geom_histogram() +
        labs(x= "Number of Regional Wins", y= "Counts", title= "Distribution of Regional Wins") + # variable being plotted is the Number of Regional Wins
        theme(text= element_text(size=12))

distribution_reg_wins
```

```r

proportion_data <- tidy_mm_data |>
                   group_by(tourney_finish) |>
                    summarize(count=n(), 
                              proportion= (n()/nrow(tidy_mm_data))) |>
                    mutate(tourney_finish= fct_recode(tourney_finish, "Top 8 Loss"="top_8_loss", "Top 4 Loss"="top_4_loss", "Top 2 Loss"="top_2_loss", 
                                                      "Top 16 Loss"="top_16_loss", "Second Round Loss"= "second_round_loss", "Opening Round Loss"= "opening_round_loss",
                                                      "First Round Loss"= "first_round_loss", "Champ"= "champ")) #renamed all the factor values to more readable names

relative_proportion_plot <- proportion_data |>
                            ggplot(aes(x= proportion, y= tourney_finish)) +
                            geom_bar(stat="identity") +
                            labs(x= "Relative Proportion", y= "Tourney Finish", title= "Relative Proportions of Tourney Finishes") + #variable plotted is the tourney finishes
                            theme(text= element_text(size=12))

relative_proportion_plot
```

Visualization 1 shows the relationship between the two predictor variables with each data point was colored by their tourney finish. This is relevant to the analysis because it helps to understand if the two predictors are correlated or not, and it reveals each data point’s tourney finish. By coloring the tourney finish of each data point, we’re able to start seeing the relationship between the two predictors and the response variable. We can begin to see whether there’s a stronger or weaker relationship between the two predictor variables and the response variable.

```r
tidy_mm_data <- tidy_mm_data |>
                mutate(tourney_finish= fct_recode(tourney_finish, "Top 8 Loss"="top_8_loss", "Top 4 Loss"="top_4_loss", "Top 2 Loss"="top_2_loss", 
                                                      "Top 16 Loss"="top_16_loss", "Second Round Loss"= "second_round_loss", "Opening Round Loss"= "opening_round_loss",
                                                      "First Round Loss"= "first_round_loss", "Champ"= "champ")) #renamed the factor values in original tidy data

mm_scatterplot <- tidy_mm_data |>
                ggplot(aes(x= reg_wins, y= reg_losses, color= tourney_finish)) +
                geom_point() +
                labs(x= "Regional Wins", y= "Regional Losses", color= "Tourney Finish", title= "Regional losses vs. Regional wins") +
                theme(text= element_text(size=12))

mm_scatterplot
```

Visualization 2 shows the relationship between regional wins and tourney finish. This is relevant to the analysis because we can see that there is a relationship between one of the predictor variables (Regional wins) and the response variable (Tourney Finish). We can see that as the tourney finish increases in rank or gets closer to Champ, the mean amount of regional wins increases. We want our predictor variable to have a clear relationship with the response variable, which is why this visualization, and the relationship it displays, is important.

```r
mm_barplot <- means_mm |>
            ggplot(aes(x= mean_reg_wins, y= fct_reorder(tourney_finish, mean_reg_wins, .desc= TRUE))) +
            geom_bar(stat= "identity") +
            labs(x= "Mean Regional Wins", y= "Tourney Finish", title= "Tourney Finish vs. Mean Regional Wins") +
            theme(text= element_text(size=12))


mm_barplot              
```
